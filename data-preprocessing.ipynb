```python
def normalize_unicode(text):
    """
    Unicode normalization without transliteration.
    """
    if text is None:
        return ""

    text = str(text)

    text = unicodedata.normalize("NFKC", text)

    return text


def normalize_text(text):
    """
    General-purpose normalization.

    - Unicode normalization
    - lowercase
    - '&' -> 'and'
    - punctuation -> spaces
    - collapse whitespace
    """

    if text is None:
        return ""

    text = normalize_unicode(text)

    text = text.lower()

    # Common semantic normalization
    text = text.replace("&", " and ")

    # Replace punctuation/symbols with spaces.
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Collapse whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
LEGAL_SUFFIXES = {
    "limited",
    "ltd",
    "llc",
    "inc",
    "incorporated",
    "corp",
    "corporation",
    "co",
    "company",
    "pvt",
    "private",
    "plc",
    "llp",
}

def normalize_business_name(text):
    text = normalize_text(text)

    if not text:
        return ""

    return text


def remove_legal_suffixes(name):
    if not name:
        return ""

    tokens = name.split()

    tokens = [
        token
        for token in tokens
        if token not in LEGAL_SUFFIXES
    ]

    return " ".join(tokens)


ADDRESS_ABBREVIATIONS = {
    "street": "st",
    "st.": "st",
    "road": "rd",
    "rd.": "rd",
    "avenue": "ave",
    "ave.": "ave",
    "boulevard": "blvd",
    "blvd.": "blvd",
    "drive": "dr",
    "dr.": "dr",
    "lane": "ln",
    "ln.": "ln",
    "highway": "hwy",
    "hwy.": "hwy",
    "parkway": "pkwy",
    "place": "pl",
    "pl.": "pl",
}


def normalize_address(text):
    text = normalize_text(text)

    if not text:
        return ""

    tokens = text.split()

    normalized_tokens = [
        ADDRESS_ABBREVIATIONS.get(token, token)
        for token in tokens
    ]

    return " ".join(normalized_tokens)


def normalize_country(value):
    if value is None:
        return ""

    value = normalize_text(value)

    if not value:
        return ""

    return COUNTRY_MAP.get(value.upper(), value.upper())


def get_tokens(text):
    if not text:
        return ""

    return " ".join(text.split())


def name_length(name):
    return len(name) if name else 0


def name_token_count(name):
    return len(name.split()) if name else 0


def address_length(address):
    return len(address) if address else 0


def address_token_count(address):
    return len(address.split()) if address else 0
```

```python
US_ZIP_PATTERN = re.compile(
    r"(?<!\d)\d{5}(?:-\d{4})?(?!\d)"
)

INDIA_PIN_PATTERN = re.compile(
    r"(?<!\d)[1-9]\d{5}(?!\d)"
)

def extract_postal_code(address, country):
    if not address:
        return ""

    country = normalize_country(country)

    if country == "US":
        match = US_ZIP_PATTERN.search(address)

        if match:
            return match.group(0)

    elif country == "IN":
        match = INDIA_PIN_PATTERN.search(address)

        if match:
            return match.group(0)

    else:
        # Generic fallback for unknown countries.
        # Keep conservative: only 5-6 digit numeric postal candidates.
        matches = re.findall(
            r"(?<!\d)\d{5,6}(?!\d)",
            address
        )

        if matches:
            return matches[-1]

    return ""
```

```python
ADDRESS_NUMBER_PATTERN = re.compile(
    r"(?<!\w)#?\d+(?:[/-]\d+)?[A-Za-z]?"
)

def extract_address_numbers(address):
    if not address:
        return ""

    numbers = ADDRESS_NUMBER_PATTERN.findall(address)

    return " ".join(numbers)


def extract_house_number(address):
    if not address:
        return ""

    # Split by commas first.
    segments = [
        seg.strip()
        for seg in address.split(",")
        if seg.strip()
    ]

    # Look for a leading address number in each segment.
    for segment in segments:

        match = re.match(
            r"^\s*#*\s*(\d+[A-Za-z]?(?:[-/]\d+)?)",
            segment
        )

        if match:
            return match.group(1)

    return ""
```

```python
ALL_US_STATE_NAMES = set(
    US_STATE_NAME_TO_CODE.keys()
)

ALL_INDIA_STATE_NAMES = set(
    INDIA_STATE_NAME_TO_CODE.keys()
)
```

```python
def split_address_segments(address):
    if not address:
        return []

    segments = [
        segment.strip()
        for segment in address.split(",")
    ]

    return [
        segment
        for segment in segments
        if segment
    ]
```

```python
def extract_us_state(address):
    """
    Extract US state using contextual address segments.

    Priority:
    1. Exact full state name in a comma-separated segment.
    2. Exact USPS 2-letter code in a comma-separated segment.
    3. State code adjacent to ZIP.
    """

    if not address:
        return "", "none"

    address_upper = normalize_unicode(address).upper()

    segments = split_address_segments(address_upper)

    # --------------------------------------------------------
    # 1. Full state name
    # --------------------------------------------------------

    for segment in segments:

        normalized_segment = re.sub(
            r"[^A-Z\s]",
            " ",
            segment
        )

        normalized_segment = re.sub(
            r"\s+",
            " ",
            normalized_segment
        ).strip()

        if normalized_segment in US_STATE_NAME_TO_CODE:
            return (
                US_STATE_NAME_TO_CODE[normalized_segment],
                "full_name_exact"
            )

    # --------------------------------------------------------
    # 2. Exact 2-letter state code in a segment
    # --------------------------------------------------------

    for segment in segments:

        clean_segment = re.sub(
            r"[^A-Z0-9\s]",
            " ",
            segment
        )

        tokens = clean_segment.split()

        for token in tokens:

            if token in US_STATE_CODES:
                return token, "state_code_token"

    # --------------------------------------------------------
    # 3. State code immediately before ZIP
    # --------------------------------------------------------

    match = re.search(
        r"(?:^|[\s,])("
        + "|".join(sorted(US_STATE_CODES, key=len, reverse=True))
        + r")(?=\s+\d{5}(?:-\d{4})?(?:\s|$))",
        address_upper
    )

    if match:
        return match.group(1), "state_before_zip"

    return "", "none"


def extract_india_state(address):
    if not address:
        return "", "none"

    address_upper = normalize_unicode(address).upper()

    segments = split_address_segments(address_upper)

    # --------------------------------------------------------
    # 1. Full state name
    # --------------------------------------------------------

    for segment in segments:

        clean_segment = re.sub(
            r"[^A-Z\s]",
            " ",
            segment
        )

        clean_segment = re.sub(
            r"\s+",
            " ",
            clean_segment
        ).strip()

        if clean_segment in INDIA_STATE_NAME_TO_CODE:
            return (
                INDIA_STATE_NAME_TO_CODE[clean_segment],
                "full_name_exact"
            )

    # --------------------------------------------------------
    # 2. Exact code in comma-separated segment
    # --------------------------------------------------------

    for segment in segments:

        clean_segment = re.sub(
            r"[^A-Z0-9\s]",
            " ",
            segment
        )

        tokens = clean_segment.split()

        for token in tokens:

            if token in INDIA_STATE_CODE_TO_NAME:
                return token, "state_code_token"

    return "", "none"


def extract_state(address, country):
    country = normalize_country(country)

    if country == "US":
        return extract_us_state(address)

    if country == "IN":
        return extract_india_state(address)

    return "", "unsupported_country"


def clean_city_candidate(segment):
    if not segment:
        return ""

    segment = normalize_text(segment)

    # Avoid treating a segment dominated by numbers as city.
    if not segment:
        return ""

    digit_count = sum(ch.isdigit() for ch in segment)
    alpha_count = sum(ch.isalpha() for ch in segment)

    if alpha_count == 0:
        return ""

    if digit_count > alpha_count:
        return ""

    return segment


def extract_city(address, country):
    """
    Heuristic city extraction.

    We only return a city when the address structure provides
    a reasonably strong candidate.

    We do not fabricate a city from external lookup data.
    """

    if not address:
        return "", "none"

    country = normalize_country(country)

    segments = split_address_segments(
        normalize_unicode(address)
    )

    if len(segments) < 2:
        return "", "none"

    # Find state first
    state_code, state_method = extract_state(
        address,
        country
    )

    if not state_code:
        return "", "no_state"

    # Determine which segment contains the state.
    state_segment_index = None

    for i, segment in enumerate(segments):

        segment_upper = segment.upper()

        # US
        if country == "US":

            clean = re.sub(
                r"[^A-Z0-9\s]",
                " ",
                segment_upper
            )

            tokens = clean.split()

            if state_code in tokens:
                state_segment_index = i
                break

        # India
        elif country == "IN":

            if state_code in segment_upper.split():
                state_segment_index = i
                break

            state_name = INDIA_STATE_CODE_TO_NAME.get(
                state_code,
                ""
            )

            if state_name and state_name in segment_upper:
                state_segment_index = i
                break

    if state_segment_index is None:
        return "", "state_position_unknown"

    candidates = []

    # Candidate before state
    if state_segment_index > 0:
        candidates.append(
            (
                state_segment_index - 1,
                segments[state_segment_index - 1],
                "before_state"
            )
        )

    # Candidate after state
    if state_segment_index + 1 < len(segments):
        candidates.append(
            (
                state_segment_index + 1,
                segments[state_segment_index + 1],
                "after_state"
            )
        )

    cleaned_candidates = []

    for idx, candidate, position in candidates:

        candidate_clean = clean_city_candidate(candidate)

        if candidate_clean:

            cleaned_candidates.append(
                (
                    idx,
                    candidate_clean,
                    position
                )
            )

    if not cleaned_candidates:
        return "", "none"

    # Prefer candidate with fewer numeric/address characteristics.
    # In practice, city segments usually contain fewer/no digits.
    for idx, candidate, position in cleaned_candidates:

        if not re.search(r"\d", candidate):
            return candidate, position

    return "", "no_confident_city"
```

```python
def preprocess_source_dataframe(df):
    """
    Add record-level normalized and derived features.

    Operates on a DataFrame containing:
        entity_id
        business_name
        business_address
        country
    """

    result = df.copy()

    # ========================================================
    # Basic normalized text
    # ========================================================

    result["name_normalized"] = (
        result["business_name"]
        .map(normalize_business_name)
    )

    result["address_normalized"] = (
        result["business_address"]
        .map(normalize_address)
    )

    result["country_normalized"] = (
        result["country"]
        .map(normalize_country)
    )

    # ========================================================
    # Tokens
    # ========================================================

    result["name_tokens"] = result["name_normalized"]

    result["address_tokens"] = result["address_normalized"]

    # ========================================================
    # Length features
    # ========================================================

    result["name_length"] = (
        result["name_normalized"].str.len()
    )

    result["name_token_count"] = (
        result["name_normalized"]
        .str.split()
        .str.len()
        .fillna(0)
        .astype("int16")
    )

    result["address_length"] = (
        result["address_normalized"].str.len()
    )

    result["address_token_count"] = (
        result["address_normalized"]
        .str.split()
        .str.len()
        .fillna(0)
        .astype("int16")
    )

    # ========================================================
    # Address numbers
    # ========================================================

    result["address_numbers"] = (
        result["address_normalized"]
        .map(extract_address_numbers)
    )

    result["house_number"] = (
        result["business_address"]
        .map(extract_house_number)
    )

    # ========================================================
    # Postal
    # ========================================================

    result["postal_code"] = [
        extract_postal_code(
            address,
            country
        )
        for address, country
        in zip(
            result["business_address"],
            result["country_normalized"]
        )
    ]

    # ========================================================
    # State
    # ========================================================

    state_results = [
        extract_state(
            address,
            country
        )
        for address, country
        in zip(
            result["business_address"],
            result["country_normalized"]
        )
    ]

    result["state_code"] = [
        x[0]
        for x in state_results
    ]

    result["state_extraction_method"] = [
        x[1]
        for x in state_results
    ]

    # ========================================================
    # City
    # ========================================================

    city_results = [
        extract_city(
            address,
            country
        )
        for address, country
        in zip(
            result["business_address"],
            result["country_normalized"]
        )
    ]

    result["city"] = [
        x[0]
        for x in city_results
    ]

    result["city_extraction_method"] = [
        x[1]
        for x in city_results
    ]

    # ========================================================
    # Validation / quality flags
    # ========================================================

    result["name_is_empty"] = (
        result["name_normalized"] == ""
    )

    result["address_is_empty"] = (
        result["address_normalized"] == ""
    )

    result["country_is_empty"] = (
        result["country_normalized"] == ""
    )

    result["state_is_empty"] = (
        result["state_code"] == ""
    )

    result["city_is_empty"] = (
        result["city"] == ""
    )

    result["postal_is_empty"] = (
        result["postal_code"] == ""
    )

    result["house_number_is_empty"] = (
        result["house_number"] == ""
    )

    return result
```

```python
def extraction_report(df, name="DATASET"):

    print("=" * 70)
    print(name)
    print("=" * 70)

    total = len(df)

    features = [
        "name_normalized",
        "address_normalized",
        "country_normalized",
        "name_tokens",
        "address_tokens",
        "house_number",
        "postal_code",
        "state_code",
        "city",
        "address_numbers",
    ]

    for col in features:

        if col not in df.columns:
            continue

        empty = (
            df[col]
            .fillna("")
            .astype(str)
            .eq("")
            .sum()
        )

        coverage = (
            100 * (total - empty) / total
            if total
            else 0
        )

        print(
            f"{col:30s} "
            f"coverage={coverage:7.2f}% "
            f"empty={empty:,}"
        )
```

```python
# ============================================================
# FINAL FEATURE SELECTION
# ============================================================

FINAL_FEATURE_COLUMNS = [
    # Identity / raw
    "entity_id",
    "business_name",
    "business_address",
    "country",

    # Name
    "name_normalized",
    "name_tokens",
    "name_length",
    "name_token_count",

    # Address
    "address_normalized",
    "address_tokens",
    "address_length",
    "address_token_count",

    # Address structure
    "address_numbers",
    "house_number",

    # Geography
    "country_normalized",
    "state_code",
    "city",
]


def select_matching_features(df):

    missing = [
        col for col in FINAL_FEATURE_COLUMNS
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Missing columns: {missing}"
        )

    return df[FINAL_FEATURE_COLUMNS].copy()


# ============================================================
# TEST
# ============================================================

test_source1_final = select_matching_features(
    test_source1_processed
)

test_source2_final = select_matching_features(
    test_source2_processed
)

test_source3_final = select_matching_features(
    test_source3_processed
)


# ============================================================
# TRAIN
# ============================================================

train_source1_final = select_matching_features(
    train_source1_processed
)

train_source2_final = select_matching_features(
    train_source2_processed
)

train_source3_final = select_matching_features(
    train_source3_processed
)


# ============================================================
# CHECK
# ============================================================

print("TEST")
print("S1:", test_source1_final.shape)
print("S2:", test_source2_final.shape)
print("S3:", test_source3_final.shape)

print("\nTRAIN")
print("S1:", train_source1_final.shape)
print("S2:", train_source2_final.shape)
print("S3:", train_source3_final.shape)

print("\nFINAL FEATURES")
for i, col in enumerate(FINAL_FEATURE_COLUMNS, 1):
    print(f"{i:2d}. {col}")
```

```python
# ============================================================
# SAVE ALL 6 DATAFRAMES
# ============================================================

DATAFRAMES_TO_SAVE = {
    "train_source1_final": train_source1_final,
    "train_source2_final": train_source2_final,
    "train_source3_final": train_source3_final,
    "test_source1_final": test_source1_final,
    "test_source2_final": test_source2_final,
    "test_source3_final": test_source3_final,
}

for name, df in DATAFRAMES_TO_SAVE.items():

    path = SAVE_DIR / f"{name}.parquet"

    print(f"\nSaving {name}")
    print(f"Shape: {df.shape}")
    print(f"Path: {path}")

    df.to_parquet(
        path,
        engine="pyarrow",
        compression="zstd",
        index=False
    )

    size_gb = path.stat().st_size / (1024 ** 3)

    print(f"Saved: {size_gb:.3f} GB")
```

```python
# ============================================================
# SAVE PROCESSING METADATA
# ============================================================

import json
from datetime import datetime

metadata = {
    "created_at": datetime.now().isoformat(),

    "feature_columns": FINAL_FEATURE_COLUMNS,

    "files": {
        name: {
            "rows": int(df.shape[0]),
            "columns": int(df.shape[1]),
            "columns_list": list(df.columns),
        }
        for name, df in DATAFRAMES_TO_SAVE.items()
    }
}

metadata_path = SAVE_DIR / "metadata.json"

with open(metadata_path, "w", encoding="utf-8") as f:
    json.dump(metadata, f, indent=2)

print("Metadata saved:", metadata_path)
```
